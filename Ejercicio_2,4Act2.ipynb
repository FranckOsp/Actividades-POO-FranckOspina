```python
import math

class Circulo:
    def __init__(self, radio: float):
        self.radio = radio

    def calcular_area(self):
        return math.pi * (self.radio ** 2)

    def calcular_perimetro(self):
        return 2 * math.pi * self.radio


class Rectangulo:
    def __init__(self, base: float, altura: float):
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return self.base * self.altura

    def calcular_perimetro(self):
        return 2 * (self.base + self.altura)


class Cuadrado:
    def __init__(self, lado: float):
        self.lado = lado

    def calcular_area(self):
        return self.lado ** 2

    def calcular_perimetro(self):
        return 4 * self.lado


class TrianguloRectangulo:
    def __init__(self, base: float, altura: float):
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return (self.base * self.altura) / 2

    def calcular_hipotenusa(self):
        return math.sqrt(self.base**2 + self.altura**2)

    def calcular_perimetro(self):
        return self.base + self.altura + self.calcular_hipotenusa()

    def determinar_tipo_triangulo(self):
        hipotenusa = self.calcular_hipotenusa()
        if self.base == self.altura == hipotenusa:
            return "Equilátero"
        elif self.base == self.altura or self.base == hipotenusa or self.altura == hipotenusa:
            return "Isósceles"
        else:
            return "Escaleno"


# --- EJERCICIOS PROPUESTOS ---

class Rombo:
    def __init__(self, diagonal_mayor: float, diagonal_menor: float, lado: float = None):
        self.diagonal_mayor = diagonal_mayor
        self.diagonal_menor = diagonal_menor
        if lado is None:
            self.lado = math.sqrt((diagonal_mayor / 2)**2 + (diagonal_menor / 2)**2)
        else:
            self.lado = lado

    def calcular_area(self):
        return (self.diagonal_mayor * self.diagonal_menor) / 2

    def calcular_perimetro(self):
        return 4 * self.lado


class Trapecio:
    def __init__(self, base_mayor: float, base_menor: float, altura: float, lado1: float, lado2: float):
        self.base_mayor = base_mayor
        self.base_menor = base_menor
        self.altura = altura
        self.lado1 = lado1
        self.lado2 = lado2

    def calcular_area(self) -> float:
        return ((self.base_mayor + self.base_menor) * self.altura) / 2

    def calcular_perimetro(self) -> float:
        return self.base_mayor + self.base_menor + self.lado1 + self.lado2

def main():
    figura1 = Circulo(radio=2)
    figura2 = Rectangulo(base=1, altura=2)
    figura3 = Cuadrado(lado=3)
    figura4 = TrianguloRectangulo(base=3, altura=5)

    #figuras propuestas:
    figura5 = Rombo(diagonal_mayor=8, diagonal_menor=6)
    figura6 = Trapecio(base_mayor=10, base_menor=6, altura=4, lado1=5, lado2=5)

    print("=== CÍRCULO ===")
    print(f"Radio: {figura1.radio} cm")
    print(f"Área: {figura1.calcular_area():.2f} cm²")
    print(f"Perímetro: {figura1.calcular_perimetro():.2f} cm\n")

    print("=== RECTÁNGULO ===")
    print(f"Base: {figura2.base} cm, Altura: {figura2.altura} cm")
    print(f"Área: {figura2.calcular_area():.2f} cm²")
    print(f"Perímetro: {figura2.calcular_perimetro():.2f} cm\n")

    print("=== CUADRADO ===")
    print(f"Lado: {figura3.lado} cm")
    print(f"Área: {figura3.calcular_area():.2f} cm²")
    print(f"Perímetro: {figura3.calcular_perimetro():.2f} cm\n")

    print("=== TRIÁNGULO RECTÁNGULO ===")
    print(f"Base: {figura4.base} cm, Altura: {figura4.altura} cm")
    print(f"Área: {figura4.calcular_area():.2f} cm²")
    print(f"Perímetro: {figura4.calcular_perimetro():.2f} cm")
    print(f"Tipo de triángulo: {figura4.determinar_tipo_triangulo()}\n")

    print("=== ROMBO (Ejercicios propuestos) ===")
    print(f"Diagonal Mayor: {figura5.diagonal_mayor} cm, Diagonal Menor: {figura5.diagonal_menor} cm")
    print(f"Lado: {figura5.lado:.2f} cm")
    print(f"Área: {figura5.calcular_area():.2f} cm²")
    print(f"Perímetro: {figura5.calcular_perimetro():.2f} cm\n")

    print("=== TRAPECIO (Ejercicios propuestos) ===")
    print(f"B. Mayor: {figura6.base_mayor} cm, B. Menor: {figura6.base_menor} cm, Altura: {figura6.altura} cm")
    print(f"Área: {figura6.calcular_area():.2f} cm²")
    print(f"Perímetro: {figura6.calcular_perimetro():.2f} cm")


if __name__ == "__main__":
    main()
```
