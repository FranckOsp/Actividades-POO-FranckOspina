```python
from enum import Enum

class TipoCuenta(Enum):
    AHORROS = "AHORROS"
    CORRIENTE = "CORRIENTE"

class CuentaBancaria:
    def __init__(self, nombres_titular: str, apellidos_titular: str,
                 numero_cuenta: int, tipo_cuenta: TipoCuenta,
                 porcentaje_interes_mensual: float = 0.0):
        self.nombres_titular = nombres_titular
        self.apellidos_titular = apellidos_titular
        self.numero_cuenta = numero_cuenta
        self.tipo_cuenta = tipo_cuenta
        self.saldo = 0.0
        self.porcentaje_interes_mensual = porcentaje_interes_mensual

    def imprimir(self):
        print(f"Nombres del titular = {self.nombres_titular}")
        print(f"Apellidos del titular = {self.apellidos_titular}")
        print(f"Número de cuenta = {self.numero_cuenta}")
        print(f"Tipo de cuenta = {self.tipo_cuenta.value}")
        print(f"Saldo = {self.saldo}")

    def consultar_saldo():
        return self.saldo

    def consignar(self, valor: float):
        if valor > 0:
            self.saldo += valor
            monto_fmt = valor
            print(f"Se ha consignado ${monto_fmt} en la cuenta. El nuevo saldo es ${self.saldo:.1f}")
        else:
            print("El valor a consignar debe ser mayor a cero.")

    def retirar(self, valor: float):
        if valor <= self.saldo:
            self.saldo -= valor
            monto_fmt = valor
            print(f"Se ha retirado ${monto_fmt} en la cuenta. El nuevo saldo es ${self.saldo:.1f}")
        else:
            print("No se puede realizar el retiro: el valor solicitado supera el saldo actual.")

    # Ejercicio propuesto: aplicar tasa de interés
    def aplicar_interes_mensual(self):
        interes = self.saldo * (self.porcentaje_interes_mensual / 100)
        self.saldo += interes
        print(f"Se aplicó el interés mensual ({self.porcentaje_interes_mensual}%). El nuevo saldo es ${self.saldo:.1f}")

if __name__ == "__main__":
    cuenta1 = CuentaBancaria("Pedro", "Pérez", 123456789, TipoCuenta.AHORROS, porcentaje_interes_mensual=1.5)
    cuenta1.imprimir()
    cuenta1.consignar(200000)
    cuenta1.consignar(300000)
    cuenta1.retirar(400000)

    # Prueba del ejercicio propuesto
    print("\n--- Aplicando interés mensual ---")
    cuenta1.aplicar_interes_mensual()
```
